```python
#Import Library
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(69)
random.seed(69)

n_asli = 100 # Data asli pas 100

# 1. List manual ben gak butuh Faker
nama_list = ['Budi', 'Ani', 'Citra', 'Doni', 'Eka', 'Fajar', 'Gina', 'Hadi', 'Indah', 'Joko',
             'Kiki', 'Lina', 'Mamat', 'Nia', 'Oki', 'Putri', 'Qori', 'Rina', 'Sari', 'Tono']
kelas_list = ['7a', '7A ', '8b', '8B', ' 9c', '9C']
gender_list = ['L', 'Laki-laki', 'P', 'Perempuan ', ' laki-laki', 'perempuan']
ekskul_list = ['Pramuka', 'pramuka ', 'Bola', 'BOLA', 'Seni', np.nan]

# Fungsi buat tanggal acak
def random_date_awur():
    start_date = datetime(2007, 1, 1) # Umur 11-18 thn dari 2025
    end_date = datetime(2014, 12, 31)
    random_days = random.randint(0, (end_date - start_date).days)
    tgl = start_date + timedelta(days=random_days)
    format_tgl = random.choice(['%Y-%m-%d', '%d/%m/%Y', '%m-%d-%Y', '%d %b %Y'])
    return tgl.strftime(format_tgl)

# 2. Bikin data asli sengaja kotor
data = {
    'nis': np.arange(1001, 1001+n_asli),
    'nama_siswa': [random.choice(nama_list).upper() if i%4==0 else random.choice(nama_list).lower() for i in range(n_asli)],
    'jenis_kelamin': np.random.choice(gender_list, n_asli),
    'umur': list(np.random.randint(11, 18, size=n_asli-3)) + [5, 150, 25], # 97 normal + 3 outlier
    'kelas': np.random.choice(kelas_list, n_asli),
    'nilai_matematika': np.random.randint(40, 100, size=n_asli).astype(float),
    'ekstrakurikuler': np.random.choice(ekskul_list, n_asli),
    'tanggal_lahir': [random_date_awur() for _ in range(n_asli)]
}

df_asli = pd.DataFrame(data)

# 3. Bikin `nilai_matematika` kosong dan & 0
df_asli.loc[np.random.choice(df_asli.index, 7, replace=False), 'nilai_matematika'] = np.nan
df_asli.loc[np.random.choice(df_asli.index, 3, replace=False), 'nilai_matematika'] = 0

# 4. BIKIN DUPLIKAT 5 BARIS saja
df_duplikat = df_asli.sample(5, random_state=1)
#df_duplikat['nis'] = df_duplikat['nis'] + 10000 # Biar NIS ne beda

# 5. Gabungkan + Acak
df = pd.concat([df_asli, df_duplikat], ignore_index=True)
df = df.sample(frac=1, random_state=2).reset_index(drop=True)

# 6. Save jadi CSV
df.to_csv('data_siswa_kotor_105.csv', index=False)
print(f"Berhasil! Total baris: {len(df)} -> 100 Asli + 5 Duplikat")
print(df.head(10))
```

```python
#Standardisasi Format Tanggal Lahir

df['tanggal_lahir'] = pd.to_datetime(
    df['tanggal_lahir'],
    format='mixed',
    errors='coerce'
)

print(df['tanggal_lahir'].dtype)
```

```python
#Standardisasi Kata Tiap Kolom

#nama_siswa Besar Awal Saja
df['nama_siswa']=df['nama_siswa'].str.lower().str.strip().str.title()

#Jenis Kelamin
#Hilangkan spasi dan ubah ke huruf kecil
df['jenis_kelamin'] = df['jenis_kelamin'].str.strip().str.lower()

# Mapping ke format yang sama
df['jenis_kelamin'] = df['jenis_kelamin'].replace({
    'l': 'Laki-laki',
    'laki-laki': 'Laki-laki',
    'p': 'Perempuan',
    'perempuan': 'Perempuan'
})

# Kelas menjadi Huruf Besar
df['kelas']=df['kelas'].str.upper().str.strip()

#Ekskul Huruf Besar Awal Saja
df['ekstrakurikuler']=df['ekstrakurikuler'].str.lower().str.strip().str.title()

```

```python
#handle missing value
#nilai mtk kosong diisi Rerata
df['nilai_matematika']=df['nilai_matematika'].fillna(df['nilai_matematika'].mean())
```
